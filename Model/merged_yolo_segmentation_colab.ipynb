```python
# 2. Install Dependencies & Check GPU
!pip install ultralytics
import ultralytics
ultralytics.checks()
```

```python
# 3. Train the Model
from ultralytics import YOLO

# Load a pre-trained YOLO segmentation model (Nano size)
model = YOLO('yolo11n-seg.pt') 

# Define the path to your new merged data.yaml
data_path = '/content/drive/MyDrive/SIH_Epilgoue/Dataset/Merged_YOLO_Dataset/data.yaml'

# Start Training!
results = model.train(
    data=data_path,
    epochs=30,            # Adjust epochs as needed
    imgsz=640,            # Image size
    batch=16,             # Batch size (reduce to 8 if you hit out-of-memory errors)
    device=0,             # Use GPU
    project='/content/drive/MyDrive/SIH_Epilgoue/Model/runs',
    name='merged_idd_rd_model'
)
```

```python
# 4. Validate the Model
# Run evaluation metrics on the validation split
metrics = model.val()
```
