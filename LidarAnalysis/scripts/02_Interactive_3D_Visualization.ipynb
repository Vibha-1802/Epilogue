# Full Interactive 3D LiDAR Visualization
This notebook is dedicated to loading and visualizing the **complete, un-downsampled** point cloud in a 3D interactive window. You will be able to explore the global structure of the LiDAR capture exactly as it was detected.

```python
import os
import open3d as o3d

print("Open3D version:", o3d.__version__)
```

## Load and Clean Data
As established previously, MATLAB exports with `float64` (`SIZE 8`), so we use the Tensor API to load the points properly. We also remove `NaN` invalid reflections to clean up the data without losing any valid points.

```python
PCD_FILE = "../lidar_data/frame_000011_t_001.000.pcd"

if not os.path.exists(PCD_FILE):
    print(f"Error: File {PCD_FILE} not found.")
else:
    # 1. Load using Tensor API (to handle SIZE 8 float64)
    t_pcd = o3d.t.io.read_point_cloud(PCD_FILE)
    
    # 2. Convert to legacy Open3D PointCloud format
    pcd = t_pcd.to_legacy()
    print(f"Points loaded: {len(pcd.points)}")
    
    # 3. Filter out NaN and Inf (invalid returns)
    pcd.remove_non_finite_points()
    print(f"Valid points after cleaning: {len(pcd.points)}")
```

## Interactive 3D Rendering (Full Resolution)
The following cell opens an external Open3D window containing **all valid points** without any downsampling. 

**Controls:**
* **Left Click & Drag:** Rotate the view
* **Right Click & Drag:** Pan the view
* **Scroll Wheel:** Zoom in and out
* **Shift + Left Click:** Change the lighting / focal point

*(Note: Depending on your system, interacting with hundreds of thousands of points at full resolution might be slightly resource-intensive).*

```python
# Visualize the full point cloud
print("Opening interactive 3D window...")

# Using draw_geometries to render the un-downsampled point cloud
o3d.visualization.draw_geometries(
    [pcd], 
    window_name=f"Full Resolution LiDAR - {os.path.basename(PCD_FILE)}",
    width=1024, 
    height=768,
    left=50,
    top=50,
    point_show_normal=False
)

print("Window closed.")
```
