```
# 3. Evaluate Model & Download Metrics
from ultralytics import YOLO
import pandas as pd
from google.colab import files

# 1. Load the trained model
# Adjust this path if you uploaded best_parinita.pt to a different location in your Drive
model_path = '/content/drive/MyDrive/SIH_Epilgoue/Model/best_parinita.pt'
model = YOLO(model_path)

# 2. Define the path to your data.yaml
data_path = '/content/drive/MyDrive/SIH_Epilgoue/Dataset/Merged_YOLO_Dataset/data.yaml'

print("Running validation... This may take a few minutes.")
# 3. Run evaluation metrics
metrics = model.val(data=data_path)

# 4. Extract and print metrics
print("\n--- Validation Metrics (Bounding Box) ---")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"mAP50: {metrics.box.map50:.4f}")
print(f"mAP75: {metrics.box.map75:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall: {metrics.box.mr:.4f}")

# If it's a segmentation model, it will also have mask metrics:
if hasattr(metrics, 'seg'):
    print("\n--- Validation Metrics (Segmentation) ---")
    print(f"mAP50-95: {metrics.seg.map:.4f}")
    print(f"mAP50: {metrics.seg.map50:.4f}")
    print(f"Precision: {metrics.seg.mp:.4f}")
    print(f"Recall: {metrics.seg.mr:.4f}")

# 5. Save metrics to CSV
results_dict = metrics.results_dict
df = pd.DataFrame([results_dict])
csv_filename = "yolo_evaluation_metrics.csv"
df.to_csv(csv_filename, index=False)
print(f"\nMetrics saved to {csv_filename}")

# 6. Download the CSV file automatically
files.download(csv_filename)
```
